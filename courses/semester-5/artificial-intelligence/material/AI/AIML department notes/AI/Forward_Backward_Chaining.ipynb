# Forward Chaining

```python
# Facts (symptoms)
facts = {
    'fever': False,
    'cough': False,
    'sore_throat': False,
    'headache': True,
    'fatigue': True,
    'rash': True,
}

# Rules (if-then statements)
rules = [
    {'if': ['fever', 'cough'], 'then': 'flu'},
    {'if': ['fever', 'sore_throat'], 'then': 'flu'},
    {'if': ['fever', 'rash'], 'then': 'measles'},
    {'if': ['headache', 'fatigue'], 'then': 'migraine'},
]
```

```python
def forward_chaining(facts, rules):
    inferred = set()
    while True:
        new_inferences = set()
        for rule in rules:
            if all(facts.get(symptom, False) for symptom in rule['if']):
                new_fact = rule['then']
                if new_fact not in facts or not facts[new_fact]:
                    new_inferences.add(new_fact)
                    facts[new_fact] = True
        if not new_inferences:
            break
        inferred.update(new_inferences)
    return inferred

# Apply forward chaining to the knowledge base
diagnosis = forward_chaining(facts, rules)
print("Diagnosis:", diagnosis)
```

# Backward Chaining

```python
# Facts (symptoms)
facts = {
    'fever': True,
    'cough': True,
    'sore_throat': True,
    'headache': True,
    'fatigue': False,
    'rash': False,
}

# Rules (if-then statements)
rules = [
    {'then': 'flu', 'if': ['fever', 'cough']},
    {'then': 'flu', 'if': ['fever', 'sore_throat']},
    {'then': 'measles', 'if': ['fever', 'rash']},
    {'then': 'migraine', 'if': ['headache', 'fatigue']},
]
```

```python
def backward_chaining(goal, facts, rules):
    if goal in facts and facts[goal]:
        return True

    for rule in rules:
        if rule['then'] == goal:
            conditions_met = all(backward_chaining(condition, facts, rules) for condition in rule['if'])
            if conditions_met:
                facts[goal] = True
                return True

    return False

# Example usage
goal = 'measles'
result = backward_chaining(goal, facts, rules)

if result:
    print(f"The goal '{goal}' is supported by the facts.")
else:
    print(f"The goal '{goal}' is not supported by the facts.")
```
